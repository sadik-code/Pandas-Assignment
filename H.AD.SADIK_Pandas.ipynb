3. How broadcasting works

Here, a has 3 rows and 1 column, while b has 1 row and 3 columns.
NumPy broadcasts the values so that both arrays become compatible for addition.

For example:
10 is added to 1, 2, 3
20 is added to 1, 2, 3
30 is added to 1, 2, 3

So the final result has 3 rows and 3 columns.
The result shape is (3, 3). This is the broadcasting case described in the assignment

4. Why is it different from two 1D arrays?

If we add two 1D arrays, for example:

np.array([10, 20, 30]) + np.array([1, 2, 3])
the result is:
[11 22 33]

But here the arrays are (3,1) and (1,3), so broadcasting creates a 3 × 3 result

Q1. Cleaning the Student Assessment Dataset

1.Load the CSV file and inspect the dataset using head(), info(), and isnull().sum()

```python
import pandas as pd
df = pd.read_csv("student_assessment_dirty.csv")
print(df.head())
print(df.info())
print(df.isnull().sum())
```

2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

```python
df["Department"] = df["Department"].replace({
    "IT": "IT",
    "it": "IT",
    "I.T.": "IT",
    "Finance": "Finance",
    "finance": "Finance"
})
```

3.Identify and remove exact duplicate student assessment records.

```python
before = len(df)
df = df.drop_duplicates()
removed_records = before - len(df)
print("Records removed:", removed_records)
```

4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately

The Marks column contains values like 88 marks, Absent, NA, blank values and 105.
First, I keep the reason for missing marks so that Absent and unentered marks are not treated as the same thing.

```python
df["Marks_Status"] = df["Marks"].astype(str).str.strip()
df["Marks_Status"] = df["Marks_Status"].replace({
    "Absent": "Absent",
    "NA": "Not entered",
    "nan": "Not entered",
    "": "Not entered"
})
```

Now convert values such as "88 marks" into numbers:

```python
df["Marks"] = pd.to_numeric(
    df["Marks"].astype(str).str.extract(r"(\d+(?:\.\d+)?)")[0],
    errors="coerce"
)
```

Now "88 marks" becomes 88, while Absent, NA and blank values become NaN.

5.Investigate missing marks and distinguish absent students from students whose marks have not been entered. State one risk of deleting missing values without investigating why they are missing.

**S109 – Sneha: NA → needs investigation

One risk of deleting missing values without investigating is that valid student records may be lost, especially when the missing value has a meaningful reason such as the student being absent.

6.Identify marks outside the valid range of 0–100 and decide how they should be handled.

```python
invalid_marks = df[(df["Marks"] < 0) | (df["Marks"] > 100)]
print(invalid_marks)
```

I would not immediately delete the student record. I would investigate the value first. For cleaning, I can mark the invalid mark as missing:

```python
df.loc[(df["Marks"] < 0) | (df["Marks"] > 100), "Marks"] = pd.NA
```
